# Neobank/BaaS Sponsor Bank Compliance - Interactive Walkthrough

## Overview & Regulatory Context

**Regulation**: OCC Third-Party Risk Management Guidance / Federal Reserve SR 11-7  
**Regulators**: OCC, Federal Reserve, FDIC  
**Key Challenge**: Charter liability for AI decisions made by neobank partners

### Critical Compliance Problem
Sponsor banks face a unique regulatory burden in the neobank/BaaS model:
- **Charter Liability**: Banks are fully liable for AI decisions made by neobank partners under their charter
- **Audit Independence**: Banks must produce complete audit evidence without relying on neobank cooperation
- **Multi-Partner Complexity**: Each neobank partner has different AI models and risk profiles
- **Examination Readiness**: OCC examiners expect immediate access to all partner decision records

### Learning Objectives
1. Track AI decisions from multiple neobank partners under a single sponsor charter
2. Create bank-controlled audit trails independent of neobank systems
3. Demonstrate charter liability documentation for regulatory examination
4. Show business continuity during neobank partner changes or unavailability

## Customer Application Generator

Create different customer profiles that neobank partners might target:

```python
def create_neobank_application(neobank_id: str, customer_profile: str) -> Dict[str, Any]:
    """
    Creates a simulated neobank customer application based on target demographics.
    """
    application_id = str(uuid.uuid4())
    
    if customer_profile == "young_professional":
        customer_data = {
            "account_application_id": application_id,
            "neobank_id": neobank_id,
            "customer_age": random.randint(25, 35),
            "annual_income": random.uniform(60000, 120000),
            "customer_state": random.choice(["CA", "NY", "TX", "FL"]),
            "employment_status": "employed",
            "account_type": "checking",
            "initial_deposit": random.uniform(500, 5000)
        }
    elif customer_profile == "student":
        customer_data = {
            "account_application_id": application_id,
            "neobank_id": neobank_id,
            "customer_age": random.randint(18, 24),
            "annual_income": random.uniform(15000, 35000),
            "customer_state": random.choice(["CA", "NY", "MA", "IL"]),
            "employment_status": "student",
            "account_type": "student_checking",
            "initial_deposit": random.uniform(50, 1000)
        }
    elif customer_profile == "gig_worker":
        customer_data = {
            "account_application_id": application_id,
            "neobank_id": neobank_id,
            "customer_age": random.randint(25, 45),
            "annual_income": random.uniform(25000, 55000),
            "customer_state": random.choice(["CA", "TX", "FL", "AZ"]),
            "employment_status": "gig_worker",
            "account_type": "gig_checking",
            "initial_deposit": random.uniform(100, 2000)
        }
    
    return customer_data

print("[SUCCESS] Customer Application Generator defined")
print("   👥 Supports different customer demographics per neobank")
print("   **Details:** Generates realistic application data")
print("   **Objective:** Aligns with each partner's target market")
```

## Neobank AI Decision Simulator

Simulate AI decisions from different neobank partners with varying model behaviors:

```python
def simulate_neobank_ai_decision(application_data: Dict[str, Any], neobank_config: Dict[str, str]) -> Dict[str, Any]:
    """
    Simulates an AI decision made by a neobank partner under the sponsor bank's charter.
    In production, this would be received from neobank AI systems via API.
    """
    neobank_id = neobank_config["neobank_id"]
    model_version = neobank_config["neobank_model_version"]
    
    # Extract customer data
    income = application_data.get("annual_income", 0)
    age = application_data.get("customer_age", 25)
    state = application_data.get("customer_state", "CA")
    
    # Base risk scoring
    risk_score = 0.5
    
    # Income-based adjustments
    if income >= 75000:
        risk_score += 0.3
    elif income >= 50000:
        risk_score += 0.15
    elif income >= 30000:
        risk_score += 0.05
    else:
        risk_score -= 0.2
    
    # Age-based adjustments
    if age >= 25 and age <= 45:
        risk_score += 0.1
    elif age < 21:
        risk_score -= 0.1
    
    # Neobank-specific model variations
    if neobank_id == "neobank-partner-alpha":
        # Alpha focuses on young professionals
        if age <= 35 and income >= 60000:
            risk_score += 0.2
        risk_score *= 1.05  # More aggressive
    elif neobank_id == "neobank-partner-beta":
        # Beta has broader demographics
        if state in ["CA", "NY", "TX"]:
            risk_score += 0.1
        risk_score *= 0.95  # More conservative
    elif neobank_id == "neobank-partner-gamma":
        # Gamma has strict requirements
        if income < 40000:
            risk_score -= 0.3
        risk_score *= 0.90  # Most conservative
    
    # Add model uncertainty
    risk_score += random.uniform(-0.1, 0.1)
    risk_score = max(0.0, min(1.0, risk_score))
    
    # Decision based on partner-specific thresholds
    approval_threshold = neobank_config.get("approval_threshold", 0.6)
    
    if risk_score >= approval_threshold:
        decision = "approve"
    elif risk_score >= 0.4:
        decision = "manual_review"
    else:
        decision = "decline"
    
    # Generate supplementary scores
    kyc_score = min(1.0, risk_score + random.uniform(0.0, 0.2))
    fraud_score = max(0.0, 1.0 - risk_score + random.uniform(-0.1, 0.1))
    credit_score = int(300 + (risk_score * 500) + random.uniform(-50, 50))
    
    return {
        "neobank_decision": decision,
        "neobank_kyc_score": round(kyc_score, 3),
        "neobank_fraud_score": round(fraud_score, 3),
        "neobank_credit_score": credit_score,
        "decision_timestamp": datetime.utcnow().isoformat(),
        "neobank_model_version": model_version,
        "bank_audit_record_id": str(uuid.uuid4()),
        "sponsor_bank_id": "SPONSOR_BANK_001",
        "exam_ready_flag": True
    }

print("[SUCCESS] Neobank AI Decision Simulator defined")
print("   [AUTOMATED] Simulates different AI model behaviors per partner")
print("   **Results:** Generates risk scores, KYC, fraud, and credit assessments")
print("   **Bank:** Links all decisions to sponsor bank charter")
print("   **Details:** Creates unique audit record IDs for tracking")
```

## Charter Liability Analysis

Analyze decision patterns across all neobank partners to understand charter risk:

```python
print("=" * 80)
print("⚖ CHARTER LIABILITY ANALYSIS")
print("=" * 80)

# Count decisions by partner and outcome
decision_stats = {}
for decision_id, partner, decision in all_decisions:
    if partner not in decision_stats:
        decision_stats[partner] = {"approve": 0, "decline": 0, "manual_review": 0, "total": 0}
    decision_stats[partner][decision] += 1
    decision_stats[partner]["total"] += 1

print("**Bank:** CHARTER LIABILITY SUMMARY:")
print("-" * 40)
total_decisions = len(all_decisions)
print(f"   Total Decisions Under Charter: {total_decisions}")
print(f"   Sponsor Bank Liability: 100% (regardless of which partner made decision)")

for partner, stats in decision_stats.items():
    partner_name = None
    partner_config = None
    
    # Find partner configuration
    for p_id, config in neobank_partners.items():
        if partner in p_id:
            partner_name = config["neobank_name"]
            partner_config = config
            break
    
    if partner_name:
        approval_rate = (stats["approve"] / stats["total"]) * 100 if stats["total"] > 0 else 0
        
        print(f"\n[MOBILE] {partner_name}:")
        print(f"   Model Version: {partner_config['neobank_model_version']}")
        print(f"   Total Decisions: {stats['total']}")
        print(f"   Approved: {stats['approve']} ({approval_rate:.1f}%)")
        print(f"   Manual Review: {stats['manual_review']}")
        print(f"   Declined: {stats['decline']}")
        print(f"   Charter Liability: 100% (sponsor bank responsible)")

print(f"\n**Objective:** REGULATORY IMPLICATIONS:")
print(f"   • Sponsor bank must defend all decisions regardless of which partner made them")
print(f"   • Model performance differences across partners are bank's responsibility")
print(f"   • Complete audit trail ensures examination readiness")
print(f"   • Partner unavailability does not affect compliance capability")
```
